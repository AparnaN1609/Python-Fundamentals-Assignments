### 1-String Concatenation

```python
user_name = 'Aparna'
greeting= 'Hello '
greeting1= greeting + " " + user_name
print(greeting1)
```

```python
greeting2= " ,welcome to Python programming"
welcome_Message= greeting1 + " " + greeting2
print(welcome_Message)
```

### 2-String Slicing and Indexing

```python
first_character = welcome_Message[0]
print(first_character)
```

```python
last_character = welcome_Message[-1]
print(last_character)
```

```python
first_five_characters = welcome_Message[:5]
print(first_five_characters)
```

```python
last_eleven_characters = welcome_Message[-11:]
print(last_eleven_characters)
```

```python
reversed_message = welcome_Message[::-1]
print(reversed_message)
```

```python
print(welcome_Message[27:33])
```

#### _or_

```python
python_start = welcome_Message.find("Python")
python_end = python_start + len("Python")
python_word = welcome_Message[python_start:python_end]
print(python_word)
```

### 3-String Methods

```python
strM = 'Python beginner tutorial'
print(strM)
```

```python
upper_Text = strM.upper()
print(upper_Text)
```

```python
lower_Text = strM.lower()
print(lower_Text)
```

```python
original_Form = strM.capitalize()
print(original_Form)
```

```python
t_Count = strM.count('t')
print(t_Count)
```

```python
replaced_Text = strM.replace("Python", "Machine Learning")
print(replaced_Text)
```

### 4-Tuples

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
t_Combine = tuple1 + tuple2
print(t_Combine)
```

```python
t_Repeated = t_Combine * 3
print(t_Repeated)
```

```python
third_Element = t_Combine[2]
print(third_Element)
```

```python
first_Three = t_Combine[:3]
print(first_Three)
```

```python
last_Three = t_Combine[-3:]
print(last_Three)
```
